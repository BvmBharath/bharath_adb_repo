```python
def fibonaaci(n):

    a,b = 0,1

    for _ in  range(n):
        print(a)
        a,b = b,a+b

fibonaaci(10)
```

```python
array = [5,3,2,4,1]

for i in range(len(array)):
    for j in range(i+1,len(array)):
        if array[i] > array[j]:
            array[i],array[j] = array[j],array[i]

print(array)
```

```python
array = [10,20,5,90,100,40]

smallest = largest = array[0]

for i in array:
    if i < smallest:
        smallest = i
    if i > largest:
        largest = i
print(smallest,largest)
```
